# Amazon Musical Instruments — meta dataset

```python
from pathlib import Path

import polars as pl

DATA_ROOT = Path("../data/")
DATA_DIR = DATA_ROOT / "amazon_musical_instruments"
ANALYZE_DIR = DATA_ROOT / "analyze"
```

## meta_Musical_Instruments 

```python
META_PATH = DATA_DIR / "meta_Musical_Instruments.jsonl"
meta = pl.scan_ndjson(META_PATH)

meta.collect_schema().names()
```

```python
main_cat = meta.select(pl.col("main_category")).collect().unique()
main_cat
```

```python
main_cat.count()
```

```python
main_cat.write_csv(ANALYZE_DIR / "main_categories.csv")
```
