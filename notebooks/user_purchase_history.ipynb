```python
import polars as pl
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
user_purchase_history = pl.read_json(
    "../data/processed/simple/user_purchase_history.json"
)
user_purchase_history.head()
```

```python
user_purchase_history = user_purchase_history.unpivot(
    variable_name="users",
    value_name="PurchaseHistory",
)
user_purchase_history.head()
```

```python
user_purchase_history_count = user_purchase_history.select(
    pl.col("PurchaseHistory").list.len().alias("PurchaseHistoryCount")
)

sns.displot(user_purchase_history_count.to_series(), bins=50, log_scale=(False, True))
plt.xlabel("number of purchases")
plt.ylabel("count (log)")
plt.show()  
```
